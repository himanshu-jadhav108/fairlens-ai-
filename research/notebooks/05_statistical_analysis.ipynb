# Notebook 05: Exploratory Statistical Testing (PROVISIONAL)

**Purpose:** Run normality tests and paired comparison diagnostics across repeated random seeds.

**Methodological Note:** All tests in this notebook are PROVISIONAL. Random seeds represent algorithmic split variance, not independent sampling observations.

**Input Files:**
- `research/results/processed/individual_runs.csv`

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
import numpy as np
from statistics.hypothesis_testing import check_normality, compute_paired_differences

runs_path = '../results/processed/individual_runs.csv'
if os.path.exists(runs_path):
    df = pd.read_csv(runs_path)
    if len(df) >= 3:
        acc_diff = df['delta_accuracy'].dropna().values
        print('Normality of Delta Accuracy:', check_normality(acc_diff))
    else:
        print('Need at least 3 runs to test.')
else:
    print('Run file not found.')
```
