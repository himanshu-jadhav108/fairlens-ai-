# Notebook 01: Benchmark Dataset Characterization & Provenance

**Purpose:** Inspect candidate dataset schemas, base rate disparities, and sensitive demographic group distributions.

**Input Files:**
- `research/datasets/` loaders
- Benchmark data (synthetic benchmark or local real datasets)

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from datasets.registry import get_dataset, list_datasets
import pandas as pd

print('Available candidate datasets:', list_datasets())
```

```python
# Inspect Adult Census Income
adult_loader = get_dataset('adult')
print('Official Title:', adult_loader.metadata.official_title)
print('Source URL:', adult_loader.metadata.source_url)
print('License:', adult_loader.metadata.license_name)
print('Sensitive Attribute:', adult_loader.metadata.sensitive_column)
print('Privileged Value:', adult_loader.metadata.privileged_value)

df_adult = adult_loader.load_data(use_synthetic_benchmark=True)
print('\nAdult shape:', df_adult.shape)
print(df_adult.groupby('sex')['income'].agg(['count', 'mean']))
```

```python
# Inspect COMPAS
compas_loader = get_dataset('compas')
df_compas = compas_loader.load_data(use_synthetic_benchmark=True)
print('COMPAS shape:', df_compas.shape)
print(df_compas.groupby('race')['two_year_recid'].agg(['count', 'mean']))
```

```python
# Inspect German Credit
german_loader = get_dataset('german')
df_german = german_loader.load_data(use_synthetic_benchmark=True)
print('German shape:', df_german.shape)
print(df_german.groupby('age_group')['credit_risk'].agg(['count', 'mean']))
```
