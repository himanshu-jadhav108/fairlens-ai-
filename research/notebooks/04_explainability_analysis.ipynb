# Notebook 04: SHAP Attribution & Explainability Stability Analysis

**Purpose:** Evaluate how bias mitigation impacts feature attribution stability using Cosine Similarity, Spearman Rank Correlation, and L1/L2 distance.

**Input Files:**
- `research/results/raw/*.json`

```python
import sys, os, glob, json
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd

raw_files = glob.glob('../results/raw/*.json')
print(f'Found {len(raw_files)} raw result files.')

rows = []
for fp in raw_files:
    with open(fp, 'r') as f:
        d = json.load(f)
    comp = d.get('attribution_comparison', {})
    rows.append({
        'experiment_id': d.get('experiment_id'),
        'dataset': d.get('dataset_name'),
        'model': d.get('model_name'),
        'mitigation': d.get('mitigation_name'),
        'cosine_sim': comp.get('cosine_similarity'),
        'spearman_rank': comp.get('spearman_rank_correlation'),
        'l1_diff': comp.get('l1_attribution_difference')
    })
df_xai = pd.DataFrame(rows)
print(df_xai.head())
```
